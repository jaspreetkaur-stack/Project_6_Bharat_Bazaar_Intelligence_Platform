# Business Question 2

## Which product categories generate the highest total sales?

This analysis helps identify the product categories that contribute the most revenue. The results can support inventory planning, marketing campaigns, and product portfolio decisions.

```python
query = """
SELECT
    Product_Category,
    ROUND(SUM(Total_Sales), 2) AS Total_Sales
FROM sales_data
GROUP BY Product_Category
ORDER BY Total_Sales DESC;
"""

category_sales = pd.read_sql(query, conn)

category_sales
```

# Query Explanation

This query groups all transactions by product category and calculates the total sales generated by each category. The results are sorted in descending order to identify the highest revenue-generating categories.

# Business Insight

After executing the query, identify:

- Which category generates the highest revenue?
- Which category contributes the least revenue?
- Is revenue concentrated in a few categories or evenly distributed?

These insights help prioritize product investments and sales strategies.

# Recommendation

The highest-performing product categories should receive greater marketing focus, inventory support, and promotional investment. Lower-performing categories should be reviewed for pricing, demand, or assortment improvements.

# Business Question 4

## How do monthly sales trends change over time?

Monthly sales trend analysis helps identify seasonal patterns, growth trends, and periods of high or low business performance. These insights support forecasting, inventory planning, and marketing decisions.

```python
query = """
SELECT
    strftime('%Y-%m', Order_Date) AS Month,
    ROUND(SUM(Total_Sales), 2) AS Total_Sales
FROM sales_data
GROUP BY Month
ORDER BY Month;
"""

monthly_sales = pd.read_sql(query, conn)

monthly_sales
```

# Query Explanation

This query extracts the year and month from the order date, groups transactions by month, and calculates total sales for each period. The results are arranged chronologically to reveal business trends over time.

# Business Insight

After reviewing the monthly sales trend, identify:

- Which month recorded the highest sales?
- Which month had the lowest sales?
- Are there seasonal or recurring sales patterns?
- Is the overall trend increasing, decreasing, or stable?

# Recommendation

Use monthly sales trends for demand forecasting, inventory planning, and promotional campaign scheduling. Months with consistently lower sales can be targeted with discounts or marketing initiatives to improve performance.

# Business Question 14

## Which product categories achieve the highest profit margin?

Profit margin measures how much profit is earned from every unit of sales. Analyzing profit margins helps identify categories that are not only generating revenue but are also highly profitable.

```python
query = """
WITH CategoryProfit AS (
    SELECT
        Product_Category,
        SUM(Total_Sales) AS Total_Sales,
        SUM(Profit) AS Total_Profit
    FROM sales_data
    GROUP BY Product_Category
)

SELECT
    Product_Category,
    ROUND(Total_Sales, 2) AS Total_Sales,
    ROUND(Total_Profit, 2) AS Total_Profit,
    ROUND((Total_Profit * 100.0) / Total_Sales, 2) AS Profit_Margin_Percent
FROM CategoryProfit
ORDER BY Profit_Margin_Percent DESC;
"""

profit_margin = pd.read_sql(query, conn)

profit_margin
```

# Query Explanation

This query uses a Common Table Expression (CTE) to first calculate total sales and total profit for each product category. It then computes the profit margin percentage using the formula:

Profit Margin (%) = (Total Profit / Total Sales) × 100

Finally, the categories are ranked by profit margin in descending order.

# Business Insight

Analyze the output to determine:

- Which product category has the highest profit margin?
- Does the category with the highest sales also have the highest profit margin?
- Which categories may require pricing or cost optimization?

Profit margin analysis provides a clearer picture of business performance than revenue alone.

# Recommendation

Focus on expanding high-margin product categories while reviewing low-margin categories for pricing improvements, supplier negotiations, or cost optimization. Balancing revenue and profitability leads to sustainable business growth.
