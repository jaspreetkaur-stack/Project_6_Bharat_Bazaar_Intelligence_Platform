# Missing Value Analysis

Missing values can reduce data quality and negatively impact business insights, SQL queries, dashboards, and machine learning models.

This section identifies missing values, measures their impact, and applies an appropriate treatment strategy.

```python
# Count missing values in each column

missing_values = df.isnull().sum()

missing_values
```

# Business Observation

The above output shows the number of missing values present in each feature. Columns with missing values require further investigation before deciding whether to remove or impute them.

```python
# Calculate missing value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage (%)": missing_percentage.round(2)
})

missing_df
```

# Business Observation

The percentage of missing values helps determine the most appropriate cleaning strategy. Features with a very small percentage of missing values can often be cleaned without significantly affecting business analysis.

```python
# Display only columns containing missing values

missing_df[missing_df["Missing Values"] > 0]
```

# Business Observation

This filtered view highlights only the columns that require attention, allowing the cleaning process to focus on relevant business attributes.

```python
# Create a backup before applying any cleaning

df_clean = df.copy()
```

# Cleaning Strategy

A backup copy of the dataset has been created before applying any cleaning operations. This ensures that the original raw dataset remains unchanged and allows the cleaning process to be reproduced safely.

```python
# Verify whether missing values exist

if missing_values.sum() == 0:
    print("✅ No missing values found in the dataset.")
else:
    print("⚠ Missing values detected.")
```

# Business Observation

The dataset has now been evaluated for missing values. If missing values are present, they will be handled using appropriate business-driven methods. If no missing values exist, no further action is required for this step.

# Duplicate Record Analysis

Duplicate records can distort business metrics such as total sales, profit, customer count, and product performance.

# Data Type Validation

Correct data types are essential for accurate business analysis. In this section, we validate and standardize the data types of all features to ensure compatibility with SQL analysis, exploratory data analysis (EDA), machine learning, and Power BI dashboards.

# Business Observation

# Business Rule Validation

Business rule validation ensures that the dataset follows logical business constraints. Invalid records can negatively affect reports, dashboards, forecasts, and business decisions.

In this section, we validate important business rules for retail transactions.

```python
# Check for invalid quantity values

invalid_quantity = df_clean[df_clean["Quantity"] <= 0]

print("Invalid Quantity Records:", len(invalid_quantity))
```

# Business Observation

Product quantity should always be greater than zero. Transactions with zero or negative quantities indicate invalid business records.

```python
# Check for invalid unit prices

invalid_price = df_clean[df_clean["Unit_Price"] <= 0]

print("Invalid Unit Price Records:", len(invalid_price))
```

# Business Observation

Products cannot have zero or negative selling prices. These records should be investigated before performing revenue analysis.

```python
# Check for invalid shipping costs

invalid_shipping = df_clean[df_clean["Shipping_Cost"] < 0]

print("Invalid Shipping Cost Records:", len(invalid_shipping))
```

# Business Observation

Shipping costs should never be negative. Negative values usually indicate data entry or system errors.

```python
# Validate discount percentage

invalid_discount = df_clean[
    (df_clean["Discount_Percent"] < 0) |
    (df_clean["Discount_Percent"] > 100)
]

print("Invalid Discount Records:", len(invalid_discount))
```

# Business Observation

Discount percentages should fall between 0% and 100%. Values outside this range are considered invalid.

```python
# Check for negative sales

invalid_sales = df_clean[df_clean["Total_Sales"] < 0]

print("Negative Sales Records:", len(invalid_sales))
```

# Business Observation

Sales revenue should not be negative. Negative sales values may indicate data inconsistencies.

```python
# Check future order dates

today = pd.Timestamp.today()

future_orders = df_clean[df_clean["Order_Date"] > today]

print("Future Order Records:", len(future_orders))
```

# Business Observation

Order dates should not occur in the future. Future-dated transactions should be verified before analysis.

```python
# Overall validation summary

validation_summary = pd.DataFrame({
    "Validation Check": [
        "Invalid Quantity",
        "Invalid Unit Price",
        "Invalid Shipping Cost",
        "Invalid Discount",
        "Negative Sales",
        "Future Dates"
    ],
    "Records Found": [
        len(invalid_quantity),
        len(invalid_price),
        len(invalid_shipping),
        len(invalid_discount),
        len(invalid_sales),
        len(future_orders)
    ]
})

validation_summary
```

# Business Validation Summary

The dataset has been evaluated against key business rules commonly used in retail analytics. These validation checks improve data reliability and ensure that future analyses, dashboards, and machine learning models are built on trustworthy data.
