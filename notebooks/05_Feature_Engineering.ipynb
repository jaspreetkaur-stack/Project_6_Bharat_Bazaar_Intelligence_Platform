# Business Feature Creation

## Objective

Create meaningful business-driven features that improve machine learning performance and provide deeper business insights.

Rather than relying only on the original variables, new features are engineered to capture customer purchasing behavior, profitability, pricing strategy, and operational efficiency.

```python
# Create a copy of the dataset
feature_df = df.copy()

feature_df.head()
```

## Feature 1 — Profit Margin (%)

Profit Margin measures the percentage of revenue retained as profit after each sale.

It is one of the most important business KPIs for profitability analysis.

```python
feature_df["Profit_Margin"] = (
    feature_df["Profit"] / feature_df["Total_Sales"]
) * 100

feature_df["Profit_Margin"] = feature_df["Profit_Margin"].round(2)

feature_df[["Profit", "Total_Sales", "Profit_Margin"]].head()
```

## Feature 2 — Revenue per Unit

Revenue per Unit measures the average revenue generated from each item sold.

This feature helps identify premium and low-value transactions.0

```python
feature_df["Revenue_Per_Unit"] = (
    feature_df["Total_Sales"] /
    feature_df["Quantity"]
).round(2)

feature_df[
    ["Quantity",
     "Total_Sales",
     "Revenue_Per_Unit"]
].head()
```

## Feature 3 — Shipping Cost Ratio

Shipping Cost Ratio measures the percentage of sales consumed by shipping expenses.

This feature evaluates logistics efficiency.

```python
feature_df["Shipping_Cost_Ratio"] = (
    feature_df["Shipping_Cost"] /
    feature_df["Total_Sales"]
) * 100

feature_df["Shipping_Cost_Ratio"] = (
    feature_df["Shipping_Cost_Ratio"]
    .round(2)
)

feature_df[
    ["Shipping_Cost",
     "Total_Sales",
     "Shipping_Cost_Ratio"]
].head()
```

## Feature 4 — High Value Order

Orders with Total Sales greater than the dataset median are classified as High Value Orders.

This feature helps distinguish premium transactions.

```python
median_sales = feature_df["Total_Sales"].median()

feature_df["High_Value_Order"] = np.where(
    feature_df["Total_Sales"] >= median_sales,
    1,
    0
)

feature_df[
    ["Total_Sales", "High_Value_Order"]
].head()
```

```python
feature_df[
    [
        "Profit_Margin",
        "Revenue_Per_Unit",
        "Shipping_Cost_Ratio",
        "High_Value_Order"
    ]
].describe()
```

# Advanced Feature Engineering

## Objective

Create business-driven categorical features that capture customer purchasing behavior, pricing strategy, and profitability.

These engineered features improve model interpretability and predictive performance.

## Feature 5 — Order Size Category

Orders are classified into Small, Medium, and Large based on quantity purchased.

```python
feature_df["Order_Size"] = pd.cut(
    feature_df["Quantity"],
    bins=[0, 3, 7, feature_df["Quantity"].max()],
    labels=["Small", "Medium", "Large"],
    include_lowest=True
)

feature_df[["Quantity", "Order_Size"]].head()
```

## Feature 6 — Price Category

Products are categorized into Budget, Standard, Premium, and Luxury based on Unit Price quartiles.

```python
feature_df["Price_Category"] = pd.qcut(
    feature_df["Unit_Price"],
    q=4,
    labels=["Budget", "Standard", "Premium", "Luxury"]
)

feature_df[
    ["Unit_Price", "Price_Category"]
].head()
```

## Feature 7 — Discount Band

Discount percentages are grouped into business-friendly ranges.

```python
feature_df["Discount_Band"] = pd.cut(
    feature_df["Discount_Percent"],
    bins=[-1, 5, 10, 20, 30],
    labels=[
        "Low",
        "Moderate",
        "High",
        "Very High"
    ]
)

feature_df[
    ["Discount_Percent", "Discount_Band"]
].head()
```

## Feature 8 — Profit Category

Orders are classified into Low, Medium, and High Profit categories based on profit distribution.

```python
feature_df["Profit_Category"] = pd.qcut(
    feature_df["Profit"],
    q=3,
    labels=[
        "Low Profit",
        "Medium Profit",
        "High Profit"
    ]
)

feature_df[
    ["Profit", "Profit_Category"]
].head()
```

```python
feature_df[
    [
        "Order_Size",
        "Price_Category",
        "Discount_Band",
        "Profit_Category"
    ]
].describe(include="all")
```

```python
categorical_features = [
    "Order_Size",
    "Price_Category",
    "Discount_Band",
    "Profit_Category"
]

for feature in categorical_features:
    print(f"\n{feature}")
    print(feature_df[feature].value_counts())
```

# Date & Time Feature Engineering

## Objective

Extract meaningful time-based features from the order date to capture seasonality, customer purchasing trends, and business cycles.

These features help machine learning models identify temporal patterns that influence sales and profitability.

## Feature 9 — Date Conversion

Convert the Order_Date column into datetime format for extracting additional calendar-based features.

```python
feature_df["Order_Date"] = pd.to_datetime(feature_df["Order_Date"])

feature_df["Order_Date"].head()
```

## Feature 10 — Year

Extract the year of each order.

```python
feature_df["Year"] = feature_df["Order_Date"].dt.year

feature_df[["Order_Date", "Year"]].head()
```

## Feature 11 — Month

Extract the month of each order to capture monthly purchasing trends.

```python
feature_df["Month"] = feature_df["Order_Date"].dt.month

feature_df[["Order_Date", "Month"]].head()
```

## Feature 12 — Quarter

Extract the financial quarter of each order.

```python
feature_df["Quarter"] = feature_df["Order_Date"].dt.quarter

feature_df[["Order_Date", "Quarter"]].head()
```

## Feature 13 — Day of Week

Extract the day of the week to analyze weekly purchasing behavior.

```python
feature_df["Day_of_Week"] = feature_df["Order_Date"].dt.day_name()

feature_df[
    ["Order_Date", "Day_of_Week"]
].head()
```

## Feature 14 — Weekend Order

Identify whether an order was placed on a weekend.

```python
feature_df["Weekend_Order"] = np.where(
    feature_df["Order_Date"].dt.weekday >= 5,
    1,
    0
)

feature_df[
    ["Order_Date", "Weekend_Order"]
].head()
```

```python
feature_df[
    [
        "Year",
        "Month",
        "Quarter",
        "Day_of_Week",
        "Weekend_Order"
    ]
].describe(include="all")
```

```python
print("Years Available:")
print(feature_df["Year"].value_counts().sort_index())

print("\nMonths Available:")
print(feature_df["Month"].value_counts().sort_index())

print("\nWeekend Orders:")
print(feature_df["Weekend_Order"].value_counts())
```

# Categorical Feature Encoding

## Objective

Machine learning algorithms require numerical input. This section converts categorical variables into numerical representations while preserving meaningful business information.

Both Label Encoding and One-Hot Encoding are applied based on the nature of the variables.

```python
# Create a copy for encoding
encoded_df = feature_df.copy()

encoded_df.head()
```

## Label Encoding

Ordinal and engineered categorical features are encoded using Label Encoding.

```python
encoded_df = feature_df.copy()

encoded_df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

label_columns = [
    "Order_Size",
    "Price_Category",
    "Discount_Band",
    "Profit_Category"
]

for col in label_columns:
    encoded_df[col] = label_encoder.fit_transform(encoded_df[col])

encoded_df[label_columns].head()
```

```python
one_hot_columns = [
    "Customer_Segment",
    "Region",
    "Country",
    "Product_Category",
    "Payment_Method",
    "Day_of_Week"
]

encoded_df = pd.get_dummies(
    encoded_df,
    columns=one_hot_columns,
    drop_first=True,
    dtype=int
)

encoded_df.head()
```

```python
print("Dataset Shape After Encoding:")
print(encoded_df.shape)
```

```python
print("Missing Values After Encoding:")
print(encoded_df.isnull().sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaled_df = encoded_df.copy()
```
